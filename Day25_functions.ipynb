```python
def test(num):
    if num == 0:
        return
    test(num - 1)
    print(num)
test(4)   
```

```python
def test(num):
    if num == 1:
        return 1
    return num + test(num - 1)
print(test(5)) 
```

```python
def count_down(n):
    if n == 0:
        return 
    print(n)
    count_down(n - 1)
count_down(3)    
```

```python
def sum(n):
    if n == 0:
        return 0
    return (n % 10) + sum(n // 10)

print(sum(123)) 
```

```python
def count_down(n):
    if n == 0:
        return 0 
    return 1 + count_down(n // 10)
print(count_down(123))
```
