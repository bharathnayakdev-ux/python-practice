```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)
print(factorial(3))
```

```python
def add(n):
    if n == 1:
        return 1
    return n + add(n - 1)
print(add(2))
```

```python
def test(n):
    if n == 0:
        return 0
    print(n)
    test(n - 1)

test(3)     
```
