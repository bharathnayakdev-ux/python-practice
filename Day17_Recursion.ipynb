```python
def count_down(n):
    if n == 0:
        return 
    print(n)
    count_down(n - 1)
    

count_down(3)    
```

```python
def count_up(n):
    if n == 0:
        return
    count_up(n - 1 ) 
    print(n)
count_up(3)    
```

```python
def sum_n(n):
    if n == 0:
        return 0
    return n + sum_n(n - 1)
print(sum_n(3))
```

```python
def test(n):
    if n == 5:
        return
    print(n)
    test(n + 1)

test(1)
```

```python
def print_n(n):
    if n == 0:
        return 0
    print(n)
    print_n(n - 1)
    
count_up(4)    
```

```python
def odd(n):
    if n == 0:
        return 
    if n % 2 != 0:
        print(n)  
    odd(n - 1)


odd(7)    




     
```
